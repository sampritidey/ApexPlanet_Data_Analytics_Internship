```python
import pandas as pd

df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset.xlsx")

print(df.head())
```

```python
print(df.shape)
```

```python
print(df.columns)
```

```python
print(df.info())
```

```python
print(df.isnull().sum())
```

```python
median_age = df['Age'].median()

print(median_age)
df['Age'] = df['Age'].fillna(median_age)
print(df['Age'].isnull().sum())
```

```python
df['City'] = df['City'].fillna('Unknown')
print(df['City'].isnull().sum())
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
print(df['Order_ID'].duplicated().sum())
```

```python
print(df['Age'].describe())
print(df['Quantity'].describe())
print(df['Unit_Price'].describe())
print(df['Total_Sales'].describe())
```

```python
calculated_sales = df['Quantity'] * df['Unit_Price']

difference = abs(calculated_sales - df['Total_Sales'])

print("Incorrect sales values:", (difference > 0.01).sum())
```

```python
print(df['Gender'].unique())
print(df['City'].unique())
print(df['Product'].unique())
print(df['Category'].unique())
```

```python
columns = ['Age', 'Quantity', 'Unit_Price', 'Total_Sales']

for column in columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df[(df[column] < lower) | (df[column] > upper)]

    print(column, "outliers:", len(outliers))
```

```python
df['Order_Date'] = pd.to_datetime(df['Order_Date'])
print(df['Order_Date'].dtype)
```

```python
df['Order_Month'] = df['Order_Date'].dt.month
df['Order_Year'] = df['Order_Date'].dt.year
df['Order_Month_Name'] = df['Order_Date'].dt.month_name()
```

```python
df.insert(
    0,
    'Transaction_ID',
    ['TXN' + str(i).zfill(4) for i in range(1, len(df) + 1)]
)
```

```python
print("Shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nDuplicate complete rows:",
      df.duplicated().sum())

print("\nData types:")
print(df.dtypes)
```

```python
print("===== FINAL DATA QUALITY CHECK =====")

print("Rows:", len(df))
print("Columns:", len(df.columns))
print("Missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate Transaction_ID:",
      df['Transaction_ID'].duplicated().sum())
print("Invalid dates:",
      df['Order_Date'].isnull().sum())

print("\n===== DATA TYPES =====")
print(df.dtypes)

print("\n===== FIRST 5 RECORDS =====")
display(df.head())
```
