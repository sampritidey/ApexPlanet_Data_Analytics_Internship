```python

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_excel("/content/ApexPlanet_Cleaned_Dataset.xlsx")
print(df.head())
```

```python
print(df.shape)
print(df.columns)
```

```python
print(df.dtypes)
```

```python
df.info()
```

```python
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:", df.duplicated().sum())
```

```python
print("\nDescriptive Statistics:")
print(df.describe())
```

```python
total_sales = df["Total_Sales"].sum()
total_quantity = df["Quantity"].sum()
average_sales = df["Total_Sales"].mean()
total_orders = df["Order_ID"].nunique()
total_customers = df["Customer_ID"].nunique()

print("\nKey Business Metrics")
print("Total Sales:", total_sales)
print("Total Quantity:", total_quantity)
print("Average Transaction Sales:", average_sales)
print("Total Orders:", total_orders)
print("Total Customers:", total_customers)
```

```python
print("\nGender Distribution:")
print(df["Gender"].value_counts())
```

```python
print("\nCity Distribution:")
print(df["City"].value_counts())
```

```python
print("\nProduct Distribution:")
print(df["Product"].value_counts())
```

```python
print("\nCategory Distribution:")
print(df["Category"].value_counts())
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Age"], bins=10)
plt.xlabel("Age")
plt.ylabel("Number of Customers")
plt.title("Customer Age Distribution")
plt.tight_layout()
plt.savefig("age_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Quantity"], bins=10)
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.title("Quantity Distribution")
plt.tight_layout()
plt.savefig("quantity_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["Total_Sales"], bins=20)
plt.xlabel("Total Sales")
plt.ylabel("Frequency")
plt.title("Total Sales Distribution")
plt.tight_layout()
plt.savefig("total_sales_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(7, 5))
df["Gender"].value_counts().plot(kind="bar")
plt.xlabel("Gender")
plt.ylabel("Number of Transactions")
plt.title("Transactions by Gender")
plt.tight_layout()
plt.savefig("gender_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
df["Category"].value_counts().plot(kind="bar")
plt.xlabel("Category")
plt.ylabel("Number of Transactions")
plt.title("Transactions by Category")
plt.tight_layout()
plt.savefig("category_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
df["Product"].value_counts().plot(kind="bar")
plt.xlabel("Product")
plt.ylabel("Number of Transactions")
plt.title("Transactions by Product")
plt.tight_layout()
plt.savefig("product_distribution.png")
plt.show()
```

```python
plt.figure(figsize=(9, 5))
df["City"].value_counts().plot(kind="bar")
plt.xlabel("City")
plt.ylabel("Number of Transactions")
plt.title("Transactions by City")
plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("city_distribution.png")
plt.show()
```

```python
correlation = df[
    ["Age", "Quantity", "Unit_Price", "Total_Sales"]
].corr()

print(correlation)
```

```python
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.tight_layout()

plt.savefig("correlation_heatmap.png")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Quantity"],
    df["Total_Sales"]
)

plt.xlabel("Quantity")
plt.ylabel("Total Sales")
plt.title("Quantity vs Total Sales")

plt.tight_layout()
plt.savefig("quantity_vs_sales.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Unit_Price"],
    df["Total_Sales"]
)

plt.xlabel("Unit Price")
plt.ylabel("Total Sales")
plt.title("Unit Price vs Total Sales")

plt.tight_layout()
plt.savefig("unit_price_vs_sales.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Age"],
    df["Total_Sales"]
)

plt.xlabel("Age")
plt.ylabel("Total Sales")
plt.title("Age vs Total Sales")

plt.tight_layout()
plt.savefig("age_vs_sales.png")
plt.show()
```

```python
monthly_sales = df.groupby(
    ["Order_Year", "Order_Month"]
)["Total_Sales"].sum().reset_index()

print(monthly_sales)
```

```python
plt.figure(figsize=(10, 5))

plt.plot(
    monthly_sales["Order_Month"].astype(str),
    monthly_sales["Total_Sales"],
    marker="o"
)

plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.title("Monthly Sales Trend")

plt.tight_layout()
plt.savefig("monthly_sales_trend.png")
plt.show()
```
